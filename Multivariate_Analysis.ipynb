```python
#Scatterplot
#This is mainly used foir multivariate integer to integer analysis...
```

```python
sns.scatterplot(x=tips['total_bill'],y=tips['tip'],hue=tips['sex'],style=tips['smoker'],size=tips['size'])
```

```python
#Bar Plot
#This is mainly used for Multivariate Integer to Categorical Analysis
```

```python
sns.barplot(x=titanic['Pclass'],y=titanic['Age'],hue=titanic['Sex'])
```

```python
#Box Plot
#This is mainly used for Multivariate Integer to Categorical Analysis
#This plot helps detect th outliers in the data
```

```python
sns.boxplot(x=titanic['Sex'],y=titanic['Age'],hue=titanic['2urvived'])
```

```python
#Dist Plot
#This is mainly used for Multivariate Integer to Categorical Analysis
#It shows frequency of data in bins or ranges...
```

```python
sns.distplot(titanic[titanic['2urvived'] == 1]['Age'],hist = False)
sns.distplot(titanic[titanic['2urvived'] == 0]['Age'],hist = False)
```

```python
#Heatmap
#This plot helps us perform analysis between Categorical and another Categorical Data
```

```python
#Crosstabs are generally used to find all possible combination between unique data in two consecutive features....
#but the problem is that when there are multiple number of unique values, the
#grid formed can be too large and hectic to analyse

#Thus we use heatmap to analyse such data in a pictorical manner...
pd.crosstab(titanic['Pclass'],titanic['2urvived'])
```

```python
sns.heatmap(pd.crosstab(titanic['2urvived'],titanic['Pclass']))
```

```python
titanic.groupby('Sex').mean()['2urvived']
```

```python
#Clustermap...same as earlier...
#These are used to observed similarity between groups...they seem to be same as heatmaps, except that they have 
#lines called dendograms to shows similarity
```

```python
sns.clustermap(pd.crosstab(titanic['sibsp'],titanic['2urvived']))
```

```python
#Pairplot
#Consider that we have to plot all numerical columns with every other numerical column....
#That would be hectic in case of multiple columns...This problem is solved using Pair Plot...

#This is used for numerical to numerical analysis...
```

```python
#Lineplot
#Same.....numerical to numerical analysis
#It is a simple scatter plot when all dots are connected...\
#Thsi is most optimally used when the x axix is time based...like day ,month or year...
```
